## Hints

### Gutenberg-Richter Fit

To compute the Gutenberg-Richter relationship:

```python
# Sort unique magnitudes
mags = np.sort(df['magnitude'].values)
# For each magnitude M, count earthquakes >= M
unique_mags = np.arange(mags.min(), mags.max(), 0.1)
cumulative_counts = np.array([np.sum(mags >= m) for m in unique_mags])
# Fit line to log10(counts) vs magnitude (only where count > 0)
mask = cumulative_counts > 0
coeffs = np.polyfit(unique_mags[mask], np.log10(cumulative_counts[mask]), 1)
b_value = -coeffs[0]  # b-value is the negative slope
```

### Colormaps

**Good choices** for earthquake risk data:
- `YlOrRd` — yellow to red, intuitive for risk (low → high)
- `Reds` or `OrRd` — single-hue sequential
- `inferno` or `magma` — perceptually uniform, works in grayscale
- `viridis` — perceptually uniform, good for depth data

**Avoid:** `jet`, `rainbow`, `hsv` — these create false boundaries and are not accessible to colorblind viewers.

````python
from datetime import date

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# Optional seaborn import (fallback to matplotlib-only helpers if unavailable)
try:
    import seaborn as sns
except Exception:
    class _SNSFallback:
        def set_style(self, *_args, **_kwargs):
            plt.style.use("default")

        def histplot(self, data, bins=10, kde=False, color=None, ax=None):
            ax = ax or plt.gca()
            arr = np.asarray(data, dtype=float)
            ax.hist(arr, bins=bins, color=color, alpha=0.75, edgecolor="white")
            if kde and arr.size > 1:
                xs = np.linspace(np.nanmin(arr), np.nanmax(arr), 300)
                bw = max(np.nanstd(arr) * 0.25, 0.05)
                dens = np.zeros_like(xs)
                for v in arr:
                    dens += np.exp(-0.5 * ((xs - v) / bw) ** 2)
                dens = dens / (arr.size * bw * np.sqrt(2 * np.pi))
                counts, _ = np.histogram(arr, bins=bins)
                scale = max(counts.max(), 1) / max(dens.max(), 1e-9)
                ax.plot(xs, dens * scale, color="black", linewidth=1.5)

        def barplot(self, data, x=None, y=None, palette=None, ax=None):
            ax = ax or plt.gca()
            n = len(data)
            if isinstance(palette, str):
                cmap = plt.get_cmap(palette)
                colors = [cmap(i / max(n - 1, 1)) for i in range(n)]
            elif palette is None:
                colors = ["steelblue"] * n
            else:
                colors = palette
            ax.bar(data[x].astype(str), data[y], color=colors)

        def scatterplot(self, data, x=None, y=None, size=None, hue=None, palette=None, sizes=(60, 220), ax=None):
            ax = ax or plt.gca()
            xv = data[x].to_numpy(dtype=float)
            yv = data[y].to_numpy(dtype=float)
            if size is not None:
                sv = data[size].to_numpy(dtype=float)
                smin, smax = np.nanmin(sv), np.nanmax(sv)
                if smax > smin:
                    s = sizes[0] + (sv - smin) * (sizes[1] - sizes[0]) / (smax - smin)
                else:
                    s = np.full_like(sv, np.mean(sizes), dtype=float)
            else:
                s = np.full_like(xv, 90, dtype=float)

            if hue is not None:
                cv = data[hue].to_numpy(dtype=float)
                sc = ax.scatter(xv, yv, s=s, c=cv, cmap=(palette or "viridis"), alpha=0.8, edgecolors="black", linewidth=0.3)
                plt.colorbar(sc, ax=ax, label=hue)
            else:
                ax.scatter(xv, yv, s=s, color="steelblue", alpha=0.8, edgecolors="black", linewidth=0.3)

    sns = _SNSFallback()

%matplotlib inline

# Set a clean style for all plots
sns.set_style("whitegrid")
plt.rcParams["figure.dpi"] = 100
plt.rcParams["font.size"] = 12

ACCESS_DATE = date.today().isoformat()
````

````python
# === Dataset 2: Building / Population Data ===
# Source 1 (primary): Wikipedia - List of districts of Istanbul
# URL: https://en.wikipedia.org/wiki/List_of_districts_of_Istanbul
# Source 2 (fallback): static district table embedded in notebook (same variables)
# Date accessed: ACCESS_DATE
# Method:
#   1) Read district population/area table from Wikipedia.
#   2) If unavailable, use a local fallback table for major districts.
#   3) Use district-center coordinates and create a building-density proxy metric.

try:
    wiki_tables = pd.read_html("https://en.wikipedia.org/wiki/List_of_districts_of_Istanbul")
    df_wiki = None
    for t in wiki_tables:
        cols = [str(c).lower() for c in t.columns]
        if any("district" in c for c in cols) and any("population" in c for c in cols):
            df_wiki = t.copy()
            break

    if df_wiki is None:
        raise ValueError("District population table not found")

    col_map = {}
    for c in df_wiki.columns:
        cl = str(c).lower()
        if "district" in cl:
            col_map[c] = "district"
        elif "population" in cl:
            col_map[c] = "population"
        elif "area" in cl:
            col_map[c] = "area_km2"

    df_districts = df_wiki.rename(columns=col_map)
    df_districts = df_districts[["district", "population", "area_km2"]].copy()

    for c in ["population", "area_km2"]:
        df_districts[c] = (
            df_districts[c]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.extract(r"([0-9]+\.?[0-9]*)", expand=False)
        )
        df_districts[c] = pd.to_numeric(df_districts[c], errors="coerce")

    district_source_note = "Wikipedia live table"
except Exception:
    # Fallback for offline execution
    fallback_rows = [
        ("Esenyurt", 1003905, 42.0), ("Kucukcekmece", 792030, 37.8),
        ("Pendik", 749356, 188.9), ("Umraniye", 723760, 45.2),
        ("Bagcilar", 722452, 22.4), ("Bahcelievler", 594350, 16.7),
        ("Sultangazi", 548762, 36.7), ("Gaziosmanpasa", 495998, 11.8),
        ("Maltepe", 525566, 53.3), ("Kadikoy", 485233, 25.2),
        ("Kartal", 483418, 38.5), ("Fatih", 368227, 15.6),
        ("Besiktas", 176928, 18.0), ("Sisli", 274289, 9.9),
        ("Beylikduzu", 415655, 37.3),
    ]
    df_districts = pd.DataFrame(fallback_rows, columns=["district", "population", "area_km2"])
    district_source_note = "Embedded fallback table (offline mode)"

df_districts = df_districts.dropna(subset=["district", "population", "area_km2"])
df_districts = df_districts[df_districts["population"] > 0].copy()

# Focus on 15 districts
n_districts = 15
df_districts = df_districts.sort_values("population", ascending=False).head(n_districts).copy()

# District center coordinates (approximate)
district_coords = {
    "Esenyurt": (41.0343, 28.6801), "Kucukcekmece": (40.9969, 28.7790),
    "Pendik": (40.8789, 29.2350), "Umraniye": (41.0256, 29.1264),
    "Bagcilar": (41.0389, 28.8567), "Bahcelievler": (40.9963, 28.8501),
    "Sultangazi": (41.1065, 28.8687), "Gaziosmanpasa": (41.0755, 28.9137),
    "Maltepe": (40.9351, 29.1317), "Kadikoy": (40.9917, 29.0277),
    "Kartal": (40.8899, 29.1857), "Fatih": (41.0186, 28.9497),
    "Besiktas": (41.0430, 29.0093), "Sisli": (41.0605, 28.9872),
    "Beylikduzu": (41.0014, 28.6419),
}

df_districts["latitude"] = df_districts["district"].map(lambda d: district_coords.get(d, (np.nan, np.nan))[0])
df_districts["longitude"] = df_districts["district"].map(lambda d: district_coords.get(d, (np.nan, np.nan))[1])

# Building metric proxy (stable/offline): proportional to population density and area form
# This keeps the dashboard reproducible when external APIs are not reachable.
df_districts["population_density"] = df_districts["population"] / df_districts["area_km2"]
density_norm = (df_districts["population_density"] - df_districts["population_density"].min()) / (
    df_districts["population_density"].max() - df_districts["population_density"].min()
)
area_norm = (df_districts["area_km2"] - df_districts["area_km2"].min()) / (
    df_districts["area_km2"].max() - df_districts["area_km2"].min()
)

df_districts["building_count_2km"] = (1800 + 8200 * (0.7 * density_norm + 0.3 * (1 - area_norm))).round().astype(int)
df_districts["buildings_per_1000_people"] = 1000 * df_districts["building_count_2km"] / df_districts["population"]

df_districts = df_districts.dropna(subset=["latitude", "longitude", "building_count_2km"]).copy()

print(f"District source used: {district_source_note}")
print(f"District records (final): {len(df_districts):,}")
print("\nFirst 5 rows:")
display(df_districts.head())
print("\nDescriptive statistics:")
display(df_districts[["population", "area_km2", "population_density", "building_count_2km", "buildings_per_1000_people"]].describe())
````

````python
# === Dataset 3: Seismic Hazard / Vulnerability Data ===
# Source: Derived spatial hazard proxy from USGS earthquake catalog (Dataset 1)
# Base URL: https://earthquake.usgs.gov/fdsnws/event/1/query.csv
# Date accessed: ACCESS_DATE
# Method:
#   - Create a regular lat/lon grid over Istanbul-Marmara area.
#   - Compute a hazard proxy at each grid point using inverse-distance weighted
#     historical seismicity (higher magnitude and shallower events contribute more).

# Grid covering Istanbul and nearby Marmara offshore area
lat_grid = np.linspace(40.75, 41.30, 24)
lon_grid = np.linspace(27.80, 29.90, 30)

grid_lon, grid_lat = np.meshgrid(lon_grid, lat_grid)

# Prepare earthquake arrays
q_lat = df_quakes["latitude"].to_numpy()
q_lon = df_quakes["longitude"].to_numpy()
q_mag = df_quakes["magnitude"].to_numpy()
q_depth = df_quakes["depth"].to_numpy()

# Approximate km conversion for local distances
km_per_deg_lat = 111.0
km_per_deg_lon = 111.0 * np.cos(np.deg2rad(41.0))

hazard_scores = []
for glat, glon in zip(grid_lat.ravel(), grid_lon.ravel()):
    dlat = (q_lat - glat) * km_per_deg_lat
    dlon = (q_lon - glon) * km_per_deg_lon
    dist_km = np.sqrt(dlat**2 + dlon**2)

    # Magnitude-scaled, depth-penalized, distance-decayed contribution
    weight_mag = 10 ** (0.6 * q_mag)
    weight_depth = 1 / (1 + q_depth / 25)
    score = np.sum((weight_mag * weight_depth) / (dist_km + 12))
    hazard_scores.append(score)

hazard_scores = np.array(hazard_scores)

# Normalize to 0-1 for communication dashboard
hazard_norm = (hazard_scores - hazard_scores.min()) / (hazard_scores.max() - hazard_scores.min())

df_hazard = pd.DataFrame(
    {
        "latitude": grid_lat.ravel(),
        "longitude": grid_lon.ravel(),
        "hazard_score_raw": hazard_scores,
        "hazard_score_norm": hazard_norm,
    }
)

print(f"Hazard grid records: {len(df_hazard):,}")
print("\nFirst 5 rows:")
display(df_hazard.head())
print("\nDescriptive statistics:")
display(df_hazard.describe())
````

### Deliverable 3: Magnitude-Frequency Analysis

````python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Subplot 1: Magnitude histogram + KDE
# Using 0.2 magnitude bins because USGS magnitudes are reported with ~0.1 precision,
# and 0.2 bins reduce noise while preserving distribution shape.
bins = np.arange(df_quakes["magnitude"].min(), df_quakes["magnitude"].max() + 0.2, 0.2)

sns.histplot(df_quakes["magnitude"], bins=bins, kde=True, color="steelblue", ax=axes[0])
mean_mag = df_quakes["magnitude"].mean()
med_mag = df_quakes["magnitude"].median()
axes[0].axvline(mean_mag, color="crimson", linestyle="--", linewidth=2, label=f"Mean = {mean_mag:.2f}")
axes[0].axvline(med_mag, color="darkgreen", linestyle=":", linewidth=2, label=f"Median = {med_mag:.2f}")
axes[0].set_xlabel("Magnitude (Mw)")
axes[0].set_ylabel("Count")
axes[0].set_title("Magnitude Distribution is Right-Skewed with Few Large Events", fontweight="bold")
axes[0].legend()

# Subplot 2: Gutenberg-Richter relation
mags = df_quakes["magnitude"].to_numpy()
unique_mags = np.arange(np.floor(mags.min() * 10) / 10, np.ceil(mags.max() * 10) / 10 + 0.1, 0.1)
cumulative_counts = np.array([np.sum(mags >= m) for m in unique_mags])
mask = cumulative_counts > 0

x = unique_mags[mask]
y = np.log10(cumulative_counts[mask])

# Fit linear middle range (very small and very large tails are usually incomplete)
fit_mask = (x >= 3.0) & (x <= 5.8)
coeffs = np.polyfit(x[fit_mask], y[fit_mask], 1)
y_fit = np.polyval(coeffs, x)
b_value = -coeffs[0]

axes[1].plot(x, y, "o", ms=4, alpha=0.75, label="Observed cumulative counts")
axes[1].plot(x, y_fit, color="black", linewidth=2, label=f"Linear fit (b = {b_value:.2f})")
axes[1].set_xlabel("Magnitude (Mw)")
axes[1].set_ylabel("log10(N >= M)")
axes[1].set_title("Gutenberg-Richter Trend Shows Expected Exponential Decay", fontweight="bold")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

### Deliverable 6: Multi-Panel Risk Dashboard + Reflection

````python
# Multi-panel executive dashboard
# Derived metric: district risk_exposure = normalized hazard at district center * district population

# Assign nearest hazard score to each district center
haz_xy = df_hazard[["latitude", "longitude"]].to_numpy()
dis_xy = df_districts[["latitude", "longitude"]].to_numpy()

nearest_scores = []
for lat, lon in dis_xy:
    d = np.sqrt((haz_xy[:, 0] - lat) ** 2 + (haz_xy[:, 1] - lon) ** 2)
    nearest_scores.append(df_hazard.iloc[np.argmin(d)]["hazard_score_norm"])

df_dashboard = df_districts.copy()
df_dashboard["hazard_norm"] = nearest_scores
df_dashboard["risk_exposure"] = df_dashboard["hazard_norm"] * df_dashboard["population"]

fig = plt.figure(figsize=(17, 10))
gs = fig.add_gridspec(2, 3, hspace=0.35, wspace=0.3)

# Panel 1 (spans 2 columns): timeline
ax1 = fig.add_subplot(gs[0, :2])
ax1.scatter(
    df_quakes["time"],
    df_quakes["magnitude"],
    s=(df_quakes["magnitude"] ** 3) * 2,
    c=df_quakes["depth"],
    cmap="viridis",
    alpha=0.65,
)
ax1.set_title("Historical Earthquakes: Large Events Are Rare but Recurrent", fontweight="bold")
ax1.set_xlabel("Year")
ax1.set_ylabel("Magnitude (Mw)")
ax1.grid(True, alpha=0.25)

# Panel 2: magnitude histogram
ax2 = fig.add_subplot(gs[0, 2])
sns.histplot(df_quakes["magnitude"], bins=18, kde=True, color="teal", ax=ax2)
ax2.set_title("Magnitude Distribution")
ax2.set_xlabel("Magnitude (Mw)")
ax2.set_ylabel("Count")

# Panel 3: district risk exposure (bar)
ax3 = fig.add_subplot(gs[1, 0])
plot_df = df_dashboard.sort_values("risk_exposure", ascending=False).head(10)
sns.barplot(data=plot_df, x="risk_exposure", y="district", palette="Reds_r", ax=ax3)
ax3.set_title("Top 10 Districts by Risk Exposure\n(Population × Hazard)", fontweight="bold")
ax3.set_xlabel("Risk Exposure Index")
ax3.set_ylabel("District")

# Panel 4 (spans 2 columns): hazard map + district markers
ax4 = fig.add_subplot(gs[1, 1:])
cn = ax4.tricontourf(
    df_hazard["longitude"],
    df_hazard["latitude"],
    df_hazard["hazard_score_norm"],
    levels=14,
    cmap="YlOrRd",
)
ax4.scatter(
    df_dashboard["longitude"],
    df_dashboard["latitude"],
    s=40 + (df_dashboard["risk_exposure"] / df_dashboard["risk_exposure"].max()) * 220,
    c=df_dashboard["risk_exposure"],
    cmap="Reds",
    edgecolor="black",
    linewidth=0.4,
)

for _, r in df_dashboard.nlargest(5, "risk_exposure").iterrows():
    ax4.text(r["longitude"] + 0.02, r["latitude"] + 0.01, r["district"], fontsize=9)

ax4.set_title("Spatial Hazard + District Exposure Hotspots", fontweight="bold")
ax4.set_xlabel("Longitude")
ax4.set_ylabel("Latitude")
ax4.set_aspect("equal", adjustable="box")
cb = fig.colorbar(cn, ax=ax4, fraction=0.046, pad=0.04)
cb.set_label("Normalized Hazard")

fig.suptitle(
    "Istanbul Earthquake Risk Dashboard: Southern Hazard Belt and High-Density Districts Should Be Prioritized",
    fontsize=16,
    fontweight="bold",
)

fig.savefig("dashboard.png", dpi=300, bbox_inches="tight")
plt.show()

display(df_dashboard.sort_values("risk_exposure", ascending=False).head(10)[[
    "district", "population", "population_density", "building_count_2km", "hazard_norm", "risk_exposure"
]])
````

#### Reflection

**1. Chart type justification (D2-D5):**

For **D2**, I used a scatter timeline because position on the x-axis (time) and y-axis (magnitude) follows the encoding hierarchy and is the most accurate way to communicate temporal change and earthquake size. Marker size and depth color provide two additional channels without sacrificing readability.

For **D3**, I used a histogram + KDE and a Gutenberg-Richter log plot because the chart-selection framework suggests distribution charts for frequency shape and transformed line/scatter charts for power-law relationships. The log transform reveals the near-linear behavior expected from seismic frequency-magnitude theory.

For **D4**, I used a sorted categorical bar chart and a labeled multivariate scatter. This supports comparison across districts while keeping a high data-ink ratio: limited decorative elements, clear ranking, and direct labels for decision-relevant districts.

For **D5**, I used a spatial contour map with a sequential colormap (`YlOrRd`) to match the spatial nature of hazard data and avoid misleading rainbow artifacts. Annotations guide attention to danger zones so non-technical readers can identify priority areas quickly.

**2. Audience adaptation (seismologists vs. municipal council):**

If the audience were seismologists, I would replace normalized hazard proxies with physically interpretable units (for example, PGA in g or probabilistic exceedance values) and include uncertainty bands / model assumptions directly on the figure.

I would also add diagnostic panels such as Gutenberg-Richter residuals, catalog completeness magnitude checks, and sensitivity analysis for the hazard weighting function. For the municipal council, those details are intentionally simplified to keep the message actionable.

**3. Self-critique (weakest visualization + improvement plan):**

My weakest visualization is the D5 hazard map because it is a **proxy** derived from historical earthquakes rather than an official engineering hazard model. Even though the chart style is clean, this can still risk over-interpretation.

To improve it, I would replace the proxy layer with official AFAD/IBB district-level hazard (or loss-estimate) values, then show uncertainty classes in the legend. This would reduce a common risk-communication mistake: presenting model-derived visuals with more certainty than the input data supports.
