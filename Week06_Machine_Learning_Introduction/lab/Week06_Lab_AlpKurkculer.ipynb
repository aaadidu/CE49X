````python
# Handle missing values
print(f'Shape before dropping missing values: {df.shape}')
print('\nMissing values per column before cleaning:')
print(df.isnull().sum())

df = df.dropna().copy()

print(f'\nShape after dropping missing values:  {df.shape}')
print('\nMissing values per column after cleaning:')
print(df.isnull().sum())
````

````python
# Separate features (X) and target (y), then train-test split
X = df.drop('target', axis=1)
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(f'X shape: {X.shape}, y shape: {y.shape}')
print(f'X_train shape: {X_train.shape}')
print(f'X_test shape:  {X_test.shape}')
print(f'y_train shape: {y_train.shape}')
print(f'y_test shape:  {y_test.shape}')

print('\nClass distribution (train):')
print(y_train.value_counts(normalize=True).sort_index())
print('\nClass distribution (test):')
print(y_test.value_counts(normalize=True).sort_index())
````

````python
# Scale features with StandardScaler
# IMPORTANT: fit on X_train only, then transform both X_train and X_test
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('Scaling complete.')
print(f'X_train_scaled shape: {X_train_scaled.shape}')
print(f'X_test_scaled shape:  {X_test_scaled.shape}')
print(f'Mean of first scaled training feature: {X_train_scaled[:, 0].mean():.4f}')
````

### Data Leakage Explanation

We must fit `StandardScaler` only on the training data so that the model learns from information that would actually be available at training time. If we fit the scaler on the full dataset before splitting, the mean and standard deviation of the test set leak into preprocessing, which makes the test set no longer truly unseen. This is a classic **data leakage** problem: performance metrics become overly optimistic because the model indirectly uses information from the evaluation data. In real deployment, we never know future patient data in advance, so train-only fitting gives a fair and realistic estimate.


````python
# Train Logistic Regression and report metrics
model_results = {}

log_reg = LogisticRegression(max_iter=1000, random_state=42)

start = time.time()
log_reg.fit(X_train_scaled, y_train)
log_reg_time = time.time() - start

log_reg_train_acc = accuracy_score(y_train, log_reg.predict(X_train_scaled))
log_reg_test_acc = accuracy_score(y_test, log_reg.predict(X_test_scaled))

model_results['Logistic Regression'] = {
    'train_accuracy': log_reg_train_acc,
    'test_accuracy': log_reg_test_acc,
    'train_time_sec': log_reg_time
}

print('Logistic Regression results:')
print(f"Training accuracy: {log_reg_train_acc:.4f}")
print(f"Test accuracy:     {log_reg_test_acc:.4f}")
print(f"Training time:     {log_reg_time:.4f} seconds")
````

````python
# Train Decision Tree and report metrics
decision_tree = DecisionTreeClassifier(random_state=42)

start = time.time()
decision_tree.fit(X_train_scaled, y_train)
decision_tree_time = time.time() - start

decision_tree_train_acc = accuracy_score(y_train, decision_tree.predict(X_train_scaled))
decision_tree_test_acc = accuracy_score(y_test, decision_tree.predict(X_test_scaled))

model_results['Decision Tree'] = {
    'train_accuracy': decision_tree_train_acc,
    'test_accuracy': decision_tree_test_acc,
    'train_time_sec': decision_tree_time
}

print('Decision Tree results:')
print(f"Training accuracy: {decision_tree_train_acc:.4f}")
print(f"Test accuracy:     {decision_tree_test_acc:.4f}")
print(f"Training time:     {decision_tree_time:.4f} seconds")

results_df = pd.DataFrame(model_results).T
display(results_df)

best_model_name = results_df['test_accuracy'].idxmax()
best_model = log_reg if best_model_name == 'Logistic Regression' else decision_tree
print(f'Best model based on test accuracy: {best_model_name}')
````

### Model Comparison

The **Logistic Regression** model performs better on the test set (about **0.833** test accuracy) compared to the Decision Tree (about **0.683** test accuracy). The Decision Tree shows clear overfitting: it reaches near-perfect training performance (~1.00) but drops significantly on test data, which indicates memorization rather than generalization. Logistic Regression has a much smaller train-test gap (~0.852 vs ~0.833), so it generalizes more reliably for unseen patients.

````python
# Part C: 5-fold cross-validation
cv_scores = cross_val_score(best_model, X_train_scaled, y_train, cv=5, scoring='f1')

print(f'F1 scores per fold: {np.round(cv_scores, 4)}')
print(f'Mean F1: {cv_scores.mean():.4f}')
print(f'Std F1:  {cv_scores.std():.4f}')
````

### Cross-Validation Interpretation

The 5-fold CV mean F1 score is around **0.796**, which is reasonably consistent with the single-split test result (strong but not perfect performance). The standard deviation (~**0.099**) suggests some fold-to-fold variability, likely due to the modest dataset size and patient heterogeneity. If variance were much higher, it would indicate the model is unstable and highly sensitive to how data is split, reducing confidence in real-world generalization.


````python
# Part D: Overfitting curve - Decision Tree with varying max_depth

depths = [1, 2, 3, 5, 8, 12, 20, None]
train_accs = []
test_accs = []

for d in depths:
    dt_depth = DecisionTreeClassifier(max_depth=d, random_state=42)
    dt_depth.fit(X_train_scaled, y_train)

    train_accs.append(dt_depth.score(X_train_scaled, y_train))
    test_accs.append(dt_depth.score(X_test_scaled, y_test))

depth_labels = [str(d) if d is not None else 'None' for d in depths]

for d, tr_acc, te_acc in zip(depth_labels, train_accs, test_accs):
    print(f'max_depth={d:>4} | train_acc={tr_acc:.4f} | test_acc={te_acc:.4f}')
````

````python
# Plot the overfitting curve: train and test accuracy vs max_depth
plt.figure(figsize=(8, 5))
plt.plot(depth_labels, train_accs, marker='o', linewidth=2, color='#4C78A8', label='Training Accuracy')
plt.plot(depth_labels, test_accs, marker='s', linewidth=2, color='#F58518', label='Test Accuracy')

plt.title('Decision Tree Overfitting Curve (max_depth vs Accuracy)')
plt.xlabel('max_depth')
plt.ylabel('Accuracy')
plt.ylim(0.6, 1.05)
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()
````

### Overfitting Curve Interpretation

Based on test accuracy, the best depth is around **`max_depth = 3`** in this split. Very shallow trees (e.g., depth 1) underfit because they are too simple to capture clinical patterns, so both train and test accuracy stay limited. As depth increases too much (e.g., 8, 12, 20, or `None`), training accuracy reaches 1.00 while test accuracy drops, which is textbook **overfitting** from the lecture: the model memorizes training data and loses generalization.
